# PanScopus — Windows automation (Research4Life → Scopus → CSV)

**Fully automated search and CSV export**, using your Windows Chrome rather than the Streamlit Cloud headless browser that Scopus blocked. **No Scopus API, no Chrome extension, no manual export.**

Open this notebook **from the PanScopus repository folder** (the same folder as `scopus_backend.py` and `run_windows_auto.py`). Then **Run All**. The notebook checks Selenium and opens Windows Chrome.

- The default advanced query is read from `sample_query.txt`; edit that file to change the query.
- Credentials are read from your Windows `R4L_USERNAME` / `R4L_PASSWORD` environment variables if set; otherwise the notebook asks privately at runtime. They are not saved in the notebook or GitHub.
- The workflow verifies the Scopus results count, selects all documents and metadata in the export dialog, then checks the actual CSV record count before reporting success.
- CSV and provenance are saved to `exports/`. This is built on the same Windows Selenium backend previously used for a verified **80/80** export.
- An authorized Research4Life/Scopus session is required. If Scopus blocks this browser, it stops rather than bypassing access controls.


```python
from pathlib import Path
import sys
import subprocess

ROOT = Path.cwd()
required = ["scopus_backend.py", "run_windows_auto.py", "sample_query.txt"]
missing = [name for name in required if not (ROOT / name).is_file()]
if missing:
    raise FileNotFoundError(
        f"Run this notebook from the PanScopus repository folder. Missing: {missing}. "
        "Use Jupyter's file browser to navigate into the extracted PanScopus-main folder."
    )

if not sys.platform.startswith("win"):
    raise RuntimeError(
        "This notebook targets Windows desktop Chrome; it will not run in Colab/Streamlit Cloud."
    )

try:
    import selenium
except ImportError:
    print("Installing free Selenium in the active Jupyter Python environment...")
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "selenium>=4.25,<5"
    ])

from run_windows_auto import run

csv_path = run()
print(f"\nVerified CSV: {csv_path}")
```
